```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
house_price_dataset=sklearn.datasets.fetch_california_housing()
```

```python
print(house_price_dataset)
```

```python
df=pd.DataFrame(house_price_dataset.data,columns=house_price_dataset.feature_names)
```

```python
df.head()
```

```python
df['price']=house_price_dataset.target
```

```python
df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.describe()
```

```python
correlation=df.corr()
```

```python
sns.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Blues')
```

```python
X=df.iloc[:, :-1]
Y=df.iloc[:, -1]
```

```python
print(X)
print(Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
model=XGBRegressor()
```

```python
model.fit(X_train,Y_train)
```

```python
train_data_prediction=model.predict(X_train)
```

```python
print(train_data_prediction)
```

```python
score_1=metrics.r2_score(Y_train,train_data_prediction)
```

```python
score_2=metrics.mean_absolute_error(Y_train,train_data_prediction)
```

```python
print("R squared error", score_1 ,":: mean absolute error", score_2)
```

```python
test_data_prediction=model.predict(X_test)
```

```python
score_3=metrics.r2_score(Y_test,test_data_prediction)
score_4=metrics.mean_absolute_error(Y_test,test_data_prediction)
print("R squared error", score_3 ,":: mean absolute error", score_4)
```

```python
plt.scatter(Y_train,train_data_prediction)
plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.title("Actual Prices vs Predicted Prices")
plt.show()
```
